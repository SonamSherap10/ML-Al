```python
import pandas as pd
```

```python
dd = pd.read_csv("datasets/bad_daraz_dataset (1).csv")
dd["price"]=pd.to_numeric(dd["price"],errors="coerce")
bad_row = (dd["price"].isna()) | (dd["price"]<0)
dd.loc[bad_row,"price"]= dd["price"].median()
```

```python
#QUARTILE DEVIATION : IQR
# Q1=dd["price"].quantile(0.25)
# Q3=dd["price"].quantile(0.75)
# Median = dd["price"].quantile(0.50)
# IQR = Q3-Q1
# low_bound = Q1 - 1.5*IQR
# high_bound = Q3 + 1.5*IQR
# outliers = dd.loc[(dd["price"]<low_bound)|(dd["price"]>high_bound),"price"]
# dd["price"]=dd["price"].clip(low_bound,high_bound) winsorization
```

```python
#STANDARD DEVIATION : Z-SCORE
mean=dd["price"].mean()
std = dd["price"].std()
dd["z_score"] = ((dd["price"]-mean)/std)  #z-score
outliers = dd[dd["z_score"].abs()>2]
outliers
```
