```python
df = pd.read_csv("datasets/student_data.csv")
```

```python
model=LinearRegression()
sd= pd.read_csv("datasets/student_marks.csv")
# # print(sd.head(5))
# # print(sd.tail(5))
# # sd.shape
# # sd.isnull().sum()

# train_data,test_data = train_test_split(
#     sd,
#     test_size = 0.20,
#     random_state=12
# )

# train_x = train_data[["StudyHours"]]
# train_y = train_data["FinalMarks"]
# model.fit(train_x,train_y)

# # # calculating weight manually 
# # x_mean = np.mean(train_x) 
# # y_mean = np.mean(train_y) 

# # w = np.sum((train_x-x_mean)*(train_y-y_mean))/ \
# #     np.sum((train_x - x_mean)**2)

# # # calculating bias manually
# # b = y_mean - w*x_mean

# # # predicting on the given data manually
# # study_hours = [2,4,6,8]

# # for hour in study_hours:
# #  prediction = w *hour+b #final marks prediction for 2 hours of study using linear regression
# #  print("predicted final marks for ",hour ,"hours studied is ",prediction)

# # testing using the segregated test data using scikit learn automatically
# X_test = test_data[["StudyHours"]]
# prediction = model.predict(X_test)
# print(prediction)
# # print(test_data[["FinalMarks"]])
```

```python
# sd.describe()
X= sd[["StudyHours","Attendance","PreviousMarks"]]
Y = sd["FinalMarks"]

X_train ,X_test,Y_train,Y_test= train_test_split(
        X,
        Y,
        random_state=15
)

model.fit(X_train,Y_train)
# for feature, coefficient in zip(X_train.columns, model.coef_):
#     print(f"{feature}: {coefficient:.4f}")


prediction = model.predict(X_test)
# pd.DataFrame({
#   "Predicted" : prediction,
#   "Actual": Y_test 
# }) 

mae = mean_absolute_error(Y_test,prediction)
mse = mean_squared_error(Y_test,prediction)

print(f"MAE : {mae}")
print(f"MSE : {mse}")
print("RMSE:", mse ** 0.5)
```

```python
X=df[[
    "study_hours",
    "attendance",
    "previous_score"
]].to_numpy()
Y=df["exam_score"].to_numpy()
X_train,X_test,Y_train,Y_test = train_test_split(
    X,
    Y,
    random_state=42,
    test_size=0.20
)
poly=PolynomialFeatures(
    include_bias= False,
    degree = 4
)
X_train_poly = poly.fit_transform(X_train)
X_test_poly= poly.transform(X_test)
model.fit(X_train_poly,Y_train)
predict = model.predict(X_test_poly)
test = pd.DataFrame({
  "Actual": Y_test,
    "Prediction" : predict
})
test
```
