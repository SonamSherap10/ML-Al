```python
# sd.describe()
X= sd[["StudyHours","Attendance","PreviousMarks"]]
Y = sd["FinalMarks"]

X_train ,X_test,Y_train,Y_test= train_test_split(
        X,
        Y,
        random_state=15
)

model.fit(X_train,Y_train)
# for feature, coefficient in zip(X_train.columns, model.coef_):
#     print(f"{feature}: {coefficient:.4f}")


prediction = model.predict(X_test)
pd.DataFrame({
  "Predicted" : prediction,
  "Actual": Y_test 
}) 
```

```python
# sd.describe()
X= sd[["StudyHours","Attendance","PreviousMarks"]].to_numpy()
Y = sd["FinalMarks"].to_numpy()

X_train ,X_test,Y_train,Y_test= train_test_split(
        X,
        Y,
        test_size=0.20,
        random_state=12
)

model.fit(X_train,Y_train)
# for feature, coefficient in zip(X_train.columns, model.coef_):
#     print(f"{feature}: {coefficient:.4f}")


prediction = model.predict(X_test)
pd.DataFrame({
  "Predicted" : prediction,
  "Actual": Y_test 
}) 
```
