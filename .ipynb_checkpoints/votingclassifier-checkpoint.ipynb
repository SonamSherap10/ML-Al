```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from  sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.metrics import (accuracy_score,classification_report)
```

```python
df = pd.read_csv("datasets/student_multiple_binary_classification.csv")

X = df[[
    "StudyHours",
    "Attendance",
    "PreviousMarks"
]]
y = df["Result"]

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Algorithms to inject in the voting classifers
lr_model = LogisticRegression(
    random_state=42
)
knn_model = KNeighborsClassifier(
    n_neighbors=5
)
dt_model = DecisionTreeClassifier(
    random_state=42
)

#  pass algorithms as args in the voting classifer 
model = VotingClassifier(
    estimators=[
        ("logistic",lr_model),
        ("knn",knn_model),
        ("dt",dt_model)
    ],
    voting= "hard"       #use either hard/soft voting 
  )

model.fit(X_train,y_train)
prediction = model.predict(X_test)
# accuracy = accuracy_score(prediction,y_test)
# print(f"The model is {accuracy*100}% accurate")

classification_eval = classification_report(y_test,prediction)
print(classification_eval)
```
