```python
from sklearn.metrics import (confusion_matrix,classification_report)
```

```python
y_test = [1,0,0,1]

y_prediction = [1,0,1,0]

cm = confusion_matrix(y_test,y_prediction)
tn,fp,fn,tp = cm.ravel()
print(f"True Negative : {tn} , False Positive : {fp} , False Negative : {fn} , True Positive : {tp}")

precision  = tp /(tp + fp)
accuracy = (tp + tn) / (tp + tn + fp + fn)
recall = tp / (tp +fn)
specificity = tn / (tn + fp)
f1 = 2 * ((precision * recall) / (precision + recall))

report = classification_report(y_test,y_prediction)

print(f"Precision : {precision} , Accuracy : {accuracy} , Recall : {recall} , Specificity : {specificity} , F1 : {f1} ")
print (f"Acutal {report}")
```
