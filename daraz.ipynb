```python
bdd = pd.read_csv("datasets/bad_daraz_dataset (1).csv")
bdd["price"]=pd.to_numeric(bdd["price"],errors="coerce") 
# bdd.fillna(bdd["price"].mean())
```

```python
bdd["price"]=bdd["price"].fillna(bdd["price"].mean())
# entering the price value by the mean average of all data
```

```python
bdd["age"]=pd.to_numeric(bdd["age"],errors="coerce")
bdd["age"]=bdd["age"].fillna(bdd["age"].median())
# entering the missing age values by median of all the age data
```
