```python
import pandas as pd 
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("datasets/student_binary_classification.csv")
df["Result"]=df["Result"].map({
    "Fail":0,
    "Good":1,
    "Average":2,
    "Excellent":3
})

X = df[["StudyHours"]] 
y = df["Result"]

model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42  #for stable and consistent model
)

model.fit(X,y)
new_student = pd.DataFrame({
    "StudyHours":[2,3,4,5,7,]
})
pd = model.predict(new_student)
pd
```
